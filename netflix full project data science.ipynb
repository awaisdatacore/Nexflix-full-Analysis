```python
# check outliers in release date column
Q1 = df['Release_Date'].quantile(0.25)
Q3 = df['Release_Date'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['Release_Date'] < lower_bound) | (df['Release_Date'] > upper_bound)]
print(outliers)
```

```python
print("Total outliers found:", len(outliers))
print(outliers['Release_Date'].value_counts())
```

```python
# Step 1: Convert to numeric (non-numeric ban jayein NaN)
df['Vote_Average'] = pd.to_numeric(df['Vote_Average'], errors='coerce')

# Step 2: Replace NaN with median
df['Vote_Average'].fillna(df['Vote_Average'].mode(), inplace=True)
```

```python
df['Vote_Count'] = df['Vote_Count'].replace('en', np.nan)
```

```python
df['Vote_Count'] = pd.to_numeric(df['Vote_Count'], errors='coerce')
```

```python
# Drop irrelevant columns
df.drop(['Overview','Original_Language','Poster_Url'], axis=1, inplace=True)
```

```python
# Fill missing values in 'genre' and 'title' with their respective modes
df['Genre'].fillna(df['Genre'].mode()[0], inplace=True)
df['Title'].fillna(df['Title'].mode()[0], inplace=True)
```

```python
df.isnull().sum()
```

```python
# Vote Average aur Popularity ko float hi rehne do
df['Popularity'].fillna(df['Popularity'].median(), inplace=True)
df['Vote_Average'].fillna(df['Vote_Average'].median(), inplace=True)
df['Vote_Count'] = df['Vote_Count'].fillna(0) 
df['Vote_Count'] = df['Vote_Count'].astype(int)


```

```python
df.isnull().sum()
```

```python
df.dtypes
```

```python
def categorize_rating(vote):
    if vote <= 5.9:
        return 'Low Rated'
    elif vote <= 7.0:
        return 'Average Rated'
    else:
        return 'High Rated'

df['rating_category'] = df['Vote_Average'].apply(categorize_rating)
```

```python
# Step 1: Remove comma and space (optional: just remove space after comma)
df['Genre'] = df['Genre'].str.replace(', ', ',')  # remove space after comma

# Step 2: Split genre into list
df['Genre'] = df['Genre'].str.split(',')

# Step 3: Explode the genre column so each genre is in a new row
df = df.explode('Genre').reset_index(drop=True)
```

```python
df['Genre'] = df['Genre'].astype('category')
```

###  Q(1) what is the most frequent genre of movies released on netflix?

```python
df['Genre'].describe()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Explode karne ke baad genre column clean ho chuka hai
# Ab catplot se visualization:
sns.catplot(
    data=df,              # tumhara DataFrame
    x="Genre",            # x-axis par genre
    kind="count",         # countplot banay ga
    height=5,             # graph ki height
    aspect=2              # width ko adjust karega
)

plt.xticks(rotation=45)   # genres ko rotate karega taake overlap na ho
plt.title("Count of Movies by Genre on Netflix")
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Step 1: Genre ki frequency count karo
genre_counts = df['Genre'].value_counts()

# Step 2: Index reset karo aur dataframe banao
genre_df = genre_counts.reset_index()
genre_df.columns = ['Genre', 'Count']

# Step 3: Sorted graph plot karo
sns.catplot(
    data=genre_df,
    x='Genre',
    y='Count',
    kind='bar',
    height=5,
    aspect=2,
    order=genre_df['Genre']  # descending order ke hisaab se
)

plt.xticks(rotation=45)
plt.title("Genres on Netflix (Sorted by Frequency)")
plt.tight_layout()
plt.show()
```

#### conclusion: the most frequent genre of movies relesed on netflix is Drama

```python
df['rating_category']
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Step 1: Count of each rating_category
rating_counts = df['rating_category'].value_counts()

# Step 2: Reset index to make a new DataFrame
rating_df = rating_counts.reset_index()
rating_df.columns = ['Rating_Category', 'Count']

# Step 3: Catplot with sorted order
sns.catplot(
    data=rating_df,
    x='Rating_Category',
    y='Count',
    kind='bar',
    height=5,
    aspect=2,
    order=rating_df['Rating_Category']  # sorted order
)

plt.title("Number of Movies per Rating Category on Netflix")
plt.xlabel("Rating Category")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### q(3) what movie got the highest popularity?what is its genre?

```python
# Step 1: Find the title of the most popular movie
most_popular_Title = df.loc[df['Popularity'].idxmax(), 'Title']

# Step 2: Filter rows with that title
filtered_df = df[df['Title'] == most_popular_Title]

# Step 3: Explode the Genre column
exploded_genres = filtered_df.explode('Genre')

# Step 4: Get unique genres
all_genres = exploded_genres['Genre'].unique()

# Step 5: Print result
print("Most Popular Movie Title:", most_popular_Title)
print("All Genres:", list(all_genres))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Step 1: Get title of most popular movie
most_popular_title = df.loc[df['Popularity'].idxmax(), 'Title']

# Step 2: Filter data for that movie only and flatten the genres
popular_movie_genres = df[df['Title'] == most_popular_title]['Genre'].explode()

# Step 3: Plot the genres
plt.figure(figsize=(6, 4))
sns.countplot(y=popular_movie_genres)
plt.title(f"Genres of Most Popular Movie:\n{most_popular_title}")
plt.xlabel("Count")
plt.ylabel("Genre")
plt.show()
```

### q(4) what movie got the lowest popularity?what is its genre?

```python
import matplotlib.pyplot as plt

# Get the row with the lowest popularity
lowest_popularity_row = df[df['Popularity'] == df['Popularity'].min()]

# Extract movie title and genre safely
movie_name = lowest_popularity_row['Title'].values[0]
movie_genre = str(lowest_popularity_row['Genre'].values[0])  # ensure string

# Split genres (in case of multiple genres)
genres = movie_genre.split(', ')

# Plot
plt.figure(figsize=(6, 4))
plt.barh(genres, [1]*len(genres), color='skyblue')
plt.title(f"Genre(s) of the Least Popular Movie: {movie_name}")
plt.xlabel("Presence")
plt.ylabel("Genre")
plt.show()
```
